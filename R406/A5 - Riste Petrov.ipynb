You will use **SymPy** to derive the maximum likelihood
estimator (MLE) for the rate parameter $\lambda$ of an exponential
distribution (`Exponential` in `sympy.stats`).

Assumptions:

- Observations $X_1, \dots, X_n$ are i.i.d. exponential with parameter $\lambda > 0$,
- For reference, the exponential density is $ f(x \mid \lambda) = \lambda e^{-\lambda x} $ for $x \ge 0$. You should be able to verify this in SymPy.
- Take $n=10$.

Use SymPy to construct the log-likelihood for $\lambda$,
take its derivative, set it equal to zero, and solve for the MLE.

```python
import sympy as sp
from sympy.stats import Exponential, E, density
```

```python
n = 10
lambda_ = sp.symbols('lambda', positive=True)
```

```python
observations = sp.symbols('x0 x1 x2 x3 x4 x5 x6 x7 x8 x9')
```

$$
f(x; \lambda) = \lambda \cdot e^{-\lambda x}, \quad x \geq 0
$$

```python
verify = Exponential('verify', lambda_)
pdf = density(verify)
pdf_expression = pdf.pdf(sp.symbols('x'))
pdf_expression
```

```python
log_likelihood = sp.Add(*[sp.log(lambda_) - lambda_ * obs for obs in observations])
```

```python
derivative = sp.diff(log_likelihood, lambda_)
```

```python
mle = sp.solve(derivative, lambda_)
mle
```
