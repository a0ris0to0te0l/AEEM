You will use **SymPy** to derive the maximum likelihood
estimator (MLE) for the rate parameter $\lambda$ of an exponential
distribution (`Exponential` in `sympy.stats`).

Assumptions:

- Observations $X_1, \dots, X_n$ are i.i.d. exponential with parameter $\lambda > 0$,
- For reference, the exponential density is $ f(x \mid \lambda) = \lambda e^{-\lambda x} $ for $x \ge 0$. You should be able to verify this in SymPy.
- Take $n=10$.

Use SymPy to construct the log-likelihood for $\lambda$,
take its derivative, set it equal to zero, and solve for the MLE.
