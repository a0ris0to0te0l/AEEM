# 1.Create a Pandas DataFrame from the data provided below.

```python
import pandas as pd

df = pd.DataFrame(rows)
```

# 2. Add a new column called unit_price_usd, defined as total revenue in USD divided by quantity.

```python
df['unit_price_usd'] = df['revenue_usd'] / df['quantity']
print (df)
```

# 3. Add a new column called weight_kg, converting weight_lb to kilograms. Use the approximate conversion 1 lb ≈ 0.454 kg.

```python
df['weight_kg'] = df['weight_lb'] * 0.454
print (df)
```

# 4. Create a filtered DataFrame that contains only transactions that satisfy both of the following conditions:

    the region is "EU" or "UK"
    the quantity is at least 10

From the filtered result, keep only the columns: date, region, product, quantity, unit_price_usd.

```python
filtered_df = df[(df['region'].isin(['EU', 'UK'])) & (df['quantity'] >= 10)][['date', 'region', 'product', 'quantity', 'unit_price_usd']]

print (filtered_df)
```

# 5. Compute the average unit price by product. The result should contain one row per product and be sorted from highest to lowest average unit price.

```python
average_unit_price = df.groupby('product')['unit_price_usd'].mean().reset_index().sort_values(by='unit_price_usd', ascending=False)

print(average_unit_price)
```

# 6. Compute the total revenue by region and product. Sort the result by total revenue in descending order.

```python
total_revenue = df.groupby(['region', 'product'])['revenue_usd'].sum().reset_index().sort_values(by='revenue_usd', ascending=False)

print(total_revenue)
```

# 7.Create a pivot table with:

    rows indexed by region
    columns given by quarter
    values equal to total revenue (summed)

```python
pivot_table = df.pivot_table(
    index='region',
    columns='quarter',
    values='revenue_usd',
    aggfunc='sum',
    fill_value=0  # Replace NaN values with 0
)
print(pivot_table)
```
