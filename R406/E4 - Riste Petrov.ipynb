Write code to generate 10, 100, 1000 observations from a normal distribution with the mean 2 & standard deviation of 1.
For each of the 3 samples generate the histogram & plot it together with the theoretical density. Use SciPy.

```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
```

```python
mean = 2
std_dev = 1

ten = np.random.normal(loc=mean, scale=std_dev, size=10)
theoretical_ten =np.random.normal(loc=0, scale=1, size=10)

hundred = np.random.normal(loc=mean, scale=std_dev, size=100)
theoretical_hundred = np.random.normal(loc=0, scale=1, size=100)

thousand = np.random.normal(loc=mean, scale=std_dev, size=1000)
theoretical_thousand = np.random.normal(loc=0, scale=1, size=1000)
```

```python
plt.hist(theoretical_ten, bins=10, density=True,  color='black')
plt.hist(ten, bins=10, density=True, color='b')
```

```python
plt.hist(theoretical_hundred, bins=30, density=True, color='black')
plt.hist(hundred, bins=30, density=True, color='b')
```

```python
plt.hist(theoretical_thousand, bins=30, density=True, color='black')
plt.hist(thousand, bins=30, density=True, color='blue')
```
