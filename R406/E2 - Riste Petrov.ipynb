# 1. Use lists instead of arrays and implement the scalar product operation (.dot()). Use one-dimesnional lists. Not allowed to use numpy.

$$ x \cdot y : = \sum_{i=1}^{n} X_i Y_i$$

```python
x = [1,2,3,4]
y = [5,6,7,8]

def scalar_prod(x, y):
    result = 0
    for xi, yi in zip(x, y):
        result += xi * yi
    return result
```

```python
scalar_prod(x,y)
```

$$ (X'X)^{-1}X'y $$

Implement this in Numpy

```python
import numpy as np

def OLS(X, Y):
    beta_hat = np.linalg.pinv(X.T @ X) @ X.T @ Y

    print(beta_hat)
```
